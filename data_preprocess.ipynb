```python
import os
import pandas as pd
import shutil
from sklearn.model_selection import train_test_split
```

```python
# Read csv
df = pd.read_csv("Brain Tumor.csv")

df['Image'] = df['Image'].astype(str) + ".jpg"

# Train-test data split
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['Class'])

def move_images(dataframe, split):
    for _, row in dataframe.iterrows():
        filename = row['Image']
        label = row['Class']

        # Create class name based on label
        if label == 1:
            class_name = "with_tumor"
        else:
            class_name = "without_tumor"

        # Define source and destination paths
        src = os.path.join("brain_tumor_dataset", filename)
        dst_dir = os.path.join("dataset_with_label", split, class_name)
        dst = os.path.join(dst_dir, filename)

        # Create directory if it doesn't exist
        os.makedirs(dst_dir, exist_ok=True)

        # Copy image to new location
        shutil.copy(src, dst)

    print(f"{split} set processed: {len(dataframe)} images")

move_images(train_df, "train")
move_images(test_df, "test")

print("All done!")
```
